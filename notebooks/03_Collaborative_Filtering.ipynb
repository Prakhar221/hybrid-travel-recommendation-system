# 03. Collaborative Filtering

Implements Singular Value Decomposition (SVD) and Matrix Factorization for rating prediction.

```python
import sys
from pathlib import Path
ROOT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT_DIR))

from src.preprocessing import DataPreprocessor
from src.collaborative import CollaborativeRecommender

preproc = DataPreprocessor()
rev_df = preproc.load_reviews()

cf = CollaborativeRecommender(algorithm='svd', n_factors=10)
cf.fit(rev_df)

user_id = rev_df['user_id'].iloc[0]
recs = cf.recommend(user_id, top_k=5)
print(f"Top Recommendations for User {user_id}:", recs)
```
